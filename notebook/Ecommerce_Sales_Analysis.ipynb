## PAHSE 03: Data Cleaning
- Data Cleaning mean Handling missing values finding and fixing or removing incorrect, broken, duplicate, or useless information in a dataset. 

### 01: Create a clean copy

```python
df_clean = df.copy()
print(df_clean.head())
```

### 02: Check and Remove Duplicate Rransaction

```python
print(df_clean.duplicated().sum())

# If duplicates exist, so let remove it 
print("Remove Duplicate Transaction")
df_clean = df_clean.drop_duplicates()

```

### 03: Now verify the duplicates

```python
print(df_clean.duplicated().sum())
```

### 04: Identify Cancelled orders

```python
df_clean["IsCancelled"] = df_clean["InvoiceNo"].astype(str).str.startswith("C")

# Now check for Cancelled Orders
print("Cancelled Orders Are: ")
print(df_clean["IsCancelled"].value_counts())
```

### 05: Check For Invalid Quntity

```python
print((df_clean["Quantity"] <= 0).sum())

# Now Inspect Invalid Qunatity
print("Invalid Quantity are: ")
print(df_clean[df_clean["Quantity"] <= 0].head())
```

### 06: Check Invalid prices

```python
print((df["UnitPrice"] <= 0).sum())

# Now inspect them
print("Invalid Prices are: ")
print(df_clean[df_clean["UnitPrice"] <= 0].head())
```

### 07: Now let fix its datatypes 
- Now let make sure the important column is the appropriate types

```python
df_clean["InvoiceNo"] = df_clean["InvoiceNo"].astype(str)
df_clean["StockCode"] = df_clean["StockCode"].astype(str)
df_clean["InvoiceDate"] = pd.to_datetime(df_clean["InvoiceDate"])
df_clean["CustomerID"] = df_clean["CustomerID"].astype("Int64")

# Now check the dataypes 
print(df_clean.dtypes)
```

### 08: Now let create the main sales Dataset
- Now we create a separate dataset containing valid sales transaction

```python
sales_df = df_clean[
    (~df_clean["IsCancelled"]) & # Means keep transaction that are not cancelled 
    (df_clean["Quantity"] > 0) & # Means keep transactions with positive quantities 
    (df_clean["UnitPrice"] > 0) # Means keep transaction with valid prices 
].copy()
```

### 09: Customer Analysis Dataset 
- When we analyze customers, then we need a valid CustomerID

```python
customer_sales_df = sales_df.dropna(subset=["CustomerID"]).copy()
```
